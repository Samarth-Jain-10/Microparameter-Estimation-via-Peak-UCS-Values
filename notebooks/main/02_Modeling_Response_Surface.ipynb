# Response Surface Modeling (Kriging)

```
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import Matern, WhiteKernel
from sklearn.metrics import r2_score, mean_squared_error
import matplotlib.pyplot as plt

print(" MODEL 1: GAUSSIAN PROCESS (KRIGING) ")
print("Training Dual-Surface Kriging Model...")

# Define Kriging Kernel
kernel = Matern(nu=1.5) + WhiteKernel(noise_level=1)
multi_gpr = GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=5, random_state=42)

# Train Model
multi_gpr.fit(X_train_scaled, y_train_scaled)

# Evaluate
y_pred_scaled = multi_gpr.predict(X_test_scaled)
r2_ucs = r2_score(y_test_scaled[:, 0], y_pred_scaled[:, 0])
r2_ym = r2_score(y_test_scaled[:, 1], y_pred_scaled[:, 1])

print(f" -> PeakUCS Testing NSE       : {r2_ucs:.4f}")
print(f" -> YoungsModulus Testing NSE : {r2_ym:.4f}")

# Plotting Results
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.scatter(y_test_scaled[:, 0], y_pred_scaled[:, 0], alpha=0.5)
plt.plot([-3, 3], [-3, 3], 'r--')
plt.title('Peak UCS: Actual vs Predicted')
plt.xlabel('Actual (Scaled)')
plt.ylabel('Predicted (Scaled)')

plt.subplot(1, 2, 2)
plt.scatter(y_test_scaled[:, 1], y_pred_scaled[:, 1], alpha=0.5, color='orange')
plt.plot([-3, 3], [-3, 3], 'r--')
plt.title('Youngs Modulus: Actual vs Predicted')
plt.xlabel('Actual (Scaled)')
plt.ylabel('Predicted (Scaled)')
plt.show()
```
